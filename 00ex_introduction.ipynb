```python
# 2) SWAP THE FUNCTION

print("2) Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable")
print("")


print("2A) Using a temporary variable")
print("")

def swap_function_tmp(x, y):
    tmp = y
    y = x
    x = tmp
    return x, y

a = 1
print("the initial value of a is:", a)
b = 2
print("the initial value of b is:", b)
a, b = swap_function_tmp(a, b)
print("the swapped value of a is:", a)
print("the swapped value of b is:", b)


print("")
print("2B) Not using a temporary variable")
print("")

def swap_function(x, y):
    return y, x
    
a = 1
print("the initial value of a is:", a)
b = 2
print("the initial value of b is:", b)
a, b = swap_function(a, b)
print("the swapped value of a is:", a)
print("the swapped value of b is:", b)
```

```python
#3) COMPUTING THE DISTANCE

print("3) Write a function that calculates and returns the euclidean distance between two points u and v, where u and v are both 2-tuples (x,y). For example, if u=(3,0) and v=(0,4), the function should return 5")

import math as m

u = (3, 0)
v = (0, 4)

def euclidian_distance_fun(a, b):
    distance = m.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2 )
    return distance

euclidian_distance = euclidian_distance_fun(u, v)

print("")
print(f"The euclidean distance between two points u={u} and v={v} is: {euclidian_distance}")
```

```python
#4) COUNTING LETTERS 

print("4) Write a program to calculate the number of times each character occurs in a given string s. Ignore differneces in capitalization")

s = "abcdeABCDE1233"

print("")
print("The characters that we want to count are:")

characters = [x for x in s]
print(characters)

print("")

print("We can make a dictionary out of it ingoring the capitalization:")

s_lower = s.lower()      # => so we ignore the capitalization 

counts = {}
for character in s_lower:
    if character in counts:          # => character in counts checks if character is one of the dictionary's keys
        counts[character] += 1       # It's already a key in the dictionary, so add 1 to its value  
    else:
        counts[character] = 1        # It’s not yet a key in the dictionary, so set its value to 1

print(counts)
```

5\. Isolating the unique

Write a function that determines and count the unique numbers in the list *l*

```python
l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]

def isolate_unique(numbers):       # => The only real input is the list
    """Return the numbers that appear exactly once in `numbers`, and how many there are."""
    counts = {}
    for i in numbers:
        if i in counts:
            counts[i] += 1
        else:
            counts[i] = 1
    find = []
    for j in counts:
        if counts[j] == 1:
            find.append(j)
    return find, len(find)

a, n = isolate_unique(l)

print(f"The numbers that are unique are: {a}")
print(f"How many unique numbers: {n}")
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
n = float(input("Set the value of the number n:"))    # => When you ask to input a number you have to use float() or int() because input() sets a string not a number
print(" ")

def fun_1(x):
    """This function returns the square of a number"""
    return x**2

def fun_2(x):
    """This function returns the cube of a number"""
    return x**3

def fun_3(x):
    """This funtion returns the number raised to the 6th power using fun_1() and fun_2()"""
    return fun_1(fun_2(x))

a = fun_1(n)
print("The square of the number you chose is:")
print(a)

print(" ")

b = fun_2(n)
print("The cube of the number you chose is:")
print(b)

print(" ")

c = fun_3(n)
print("The 6th power of the number you chose is:")
print(c)
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
# To avoid duplicates we require an order: a < b < c
#  - y starts at x+1 to avoid swapped duplicates like (4, 3, 5)
#  - z starts at y+1 because the hypotenuse is always the longest side
pythagorean_list = [(x, y, z) 
                    for x in range(1, 100) 
                    for y in range(x + 1, 100) 
                    for z in range(y + 1, 100)
                    if x**2 + y ** 2 == z**2   ]   

pythagorean_tupla = tuple(pythagorean_list)
print(pythagorean_tupla)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
import math

def normalized(t):
    """Return the tuple t scaled so that its Euclidean norm is 1."""
    N = math.sqrt(sum( i ** 2 for i in t))
    return tuple( i/N for i in t)

tupla_of_num = (1, 2, 3) 

print(normalized(tupla_of_num))
```
