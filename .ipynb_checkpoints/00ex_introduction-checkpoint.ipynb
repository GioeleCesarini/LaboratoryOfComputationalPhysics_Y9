```python
# 1) MICKEYMOUSE PROBLEM

#####
## A
#####
s="Write a program that prints the numbers from 1 to 100. \
But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
For numbers which are multiples of both three and five print MickeyMouse"

print("1A) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print Mickey instead of the corresponding number and for the multiples of 5 print Mouse. For numbers which are multiples of both three and five print MickeyMouse")
print("")

lista = []

for i in range(1, 101):
    if i%3 == 0:                                        # Where: "%" is the "remainder of integer division"
        if i%5 == 0:
            print("MickeyMouse")
            lista.append("MickeyMouse")
        else:
            print("Mickey")
            lista.append("Mickey")
    elif i%5 == 0:
        print("Mouse")
        lista.append("Mouse")
    else:
        print(i)
        lista.append(i)

#####
## B    => Place the result of Part A into a tuple. To do this I’ll put it into a list first and then convert it into a tuple and then replace [the contents] with "Donald" and "Duck".
#####

print("")
print("1B) Put the result in a tuple and substitute Mickey with Donald and Mouse with Duck")
print("")

lista_nuova = []

for x in lista:
    if x == "Mickey":
        lista_nuova.append("Donald")
    elif x == "Mouse":
        lista_nuova.append("Duck")
    else:
        lista_nuova.append(x)

tupla_nuova = tuple(lista_nuova)

print(tupla_nuova)
```

```python
# 2) SWAP THE FUNCTION

print("2) Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable")
print("")


print("2A) Using a temporary variable")
print("")

def swap_function_tmp(x, y):
    tmp = y
    y = x
    x = tmp
    return x, y

a = 1
print("the initial value of a is:", a)
b = 2
print("the initial value of b is:", b)
a, b = swap_function_tmp(a, b)
print("the swapped value of a is:", a)
print("the swapped value of b is:", b)


print("")
print("2B) Not using a temporary variable")
print("")

def swap_function(x, y):
    return y, x
    
a = 1
print("the initial value of a is:", a)
b = 2
print("the initial value of b is:", b)
a, b = swap_function(a, b)
print("the swapped value of a is:", a)
print("the swapped value of b is:", b)
```

```python
#3) COMPUTING THE DISTANCE

print("3) Write a function that calculates and returns the euclidean distance between two points u and v, where u and v are both 2-tuples (x,y). For example, if u=(3,0) and v=(0,4), the function should return 5")

import math as m

u = (3, 0)
v = (0, 4)

def euclidian_distance_fun(a, b):
    distance = m.sqrt((a[0] - b[0])**2 + (a[1] - b[1])**2 )
    return distance

euclidian_distance = euclidian_distance_fun(u, v)

print("")
print(f"The euclidean distance between two points u={u} and v={v} is: {euclidian_distance}")
```
